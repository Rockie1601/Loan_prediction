```python
def correlation(dataset,threshold):
    cor_col=set()
    cor_matrix=dataset.corr()
    for i in range(len(cor_matrix.columns)):
        for j in range (i):
            if abs(cor_matrix.iloc[i,j])>threshold:
                colname=cor_matrix.columns[i]
                cor_col.add(colname)

    return cor_col
```

```python
col=correlation(df_corr,0.85)
print(col)
```

```python
from sklearn.model_selection import train_test_split
X=df.iloc[:,:-1]
y=df.iloc[:,-1]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
X_train.shape,y_train.shape
```

```python
colu = ['person_gender', 'person_education', 'person_home_ownership', 'loan_intent', 'previous_loan_defaults_on_file']
df[colu].apply(lambda col: col.value_counts())
```

```python
from sklearn.preprocessing import LabelEncoder
l1=LabelEncoder()
l2=LabelEncoder()
X_train['person_gender']=l1.fit_transform(X_train['person_gender'])
X_train['previous_loan_defaults_on_file']=l2.fit_transform(X_train['previous_loan_defaults_on_file'])
X_test['person_gender']=l1.transform(X_test['person_gender'])
X_test['previous_loan_defaults_on_file']=l2.transform(X_test['previous_loan_defaults_on_file'])
```

```python
X_train
```

```python
X_test
```

```python
X_train.drop(col,axis=1,inplace=True)
X_test.drop(col,axis=1,inplace=True)
```

```python
X_train.shape,X_test.shape
```

```python
X_train.head()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct=ColumnTransformer(transformers=[('onehotencoding', OneHotEncoder(),[2,4,6])],remainder='passthrough')
X_train=ct.fit_transform(X_train)
X_test=ct.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
lr=LogisticRegression()
param={'penalty':['l1', 'l2', 'elasticnet'],
      'C':[0.01,0.1,1,10,100],
      'solver':['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']}
grid=GridSearchCV(lr,param)
grid.fit(X_train,y_train)
```

```python
y_pred=grid.predict(X_test)
print(y_pred)
```

```python
lr.fit(X_train,y_train)
```

```python
lr.score(X_train,y_train)
```

```python
grid.score(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
as_grid=accuracy_score(y_test,y_pred)
print(as_grid)
```

```python
y_pred_lr=lr.predict(X_test)
```

```python
as_lr=accuracy_score(y_test,y_pred_lr)
print(as_lr)
```
